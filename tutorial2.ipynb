```python
#2. Create a random numpy array of size 3 x 4

#The first number is the row while the seccond is the column
#So question asks for an array of 3 rows and 4 columns
#np.random.rand generates a random array, the numbers generated will be in the form of float

array_3x4 = np.random.rand (3,4)

print ("This is my 3x4 array")
print (array_3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
#numpy only has ones and zeros, no such thing is twos

ones_array = np.ones ((4,4))
zeros_array = np.zeros ((4,4))


print ( "Here is my array of ones :\n")
print (ones_array)

print ("\n\nHere is my array of zeros :\n")
print (zeros_array)

```
