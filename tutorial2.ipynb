```python
#Part 1-Google Colab and Python
#1. Create a new notebook and code the following
#1. Create a numpy array of size 2 x 2

#numpy is a python library that allows you to work with numbers and large datasets etc.

import numpy as np

array_2x2 = np.array ([

    [1,2],
    [3,4]

])

print ( "This is my 2x2 array")
print ( array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

#The first number is the row while the seccond is the column
#So question asks for an array of 3 rows and 4 columns
#np.random.rand generates a random array, the numbers generated will be in the form of float

array_3x4 = np.random.rand (3,4)

print ("This is my 3x4 array")
print (array_3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
#numpy only has ones and zeros, no such thing is twos

ones_array = np.ones ((4,4))
zeros_array = np.zeros ((4,4))


print ( "Here is my array of ones :\n")
print (ones_array)

print ("\n\nHere is my array of zeros :\n")
print (zeros_array)

```

```python
#4. Create a random numpy array of size 6x3 and transpose it
#Transposing uses an attribute labeled .T

array_6x3 = np.random.rand(6,3)
print ("Original array :\n")
print (array_6x3)

array_6x3_transposed = array_6x3.T
print ("\nTransposed array :\n")
print (array_6x3_transposed)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

n = 3
p = 4
m = 2

array_np = np.random.rand (n,p)
array_pm = np.random.rand (p,m)


print ("First array :")
print (array_np)

print ("\nSecond array :")
print (array_pm)

#first method of representing dot product arrays
dot_product_array = np.dot (array_np, array_pm)
print ("\n The dot product of both arrays :")
print (dot_product_array)

#second method of representing dot product arrays
print ("\nThe dot product (again) : ")
print (array_np @ array_pm)


#If you want to know the shape of shape of your dot product use the term .shape
print ("\nThe shape of the dot product is :")
print (dot_product_array.shape)
```
